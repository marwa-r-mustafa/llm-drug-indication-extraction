````python
import json
import time
import requests

MODEL = "koesn/llama3-openbiollm-8b:q5_K_M"
URL = "http://127.0.0.1:11434/api/generate"

SYSTEM_MESSAGE = """
You are a biomedical information extraction system.
Follow the user's extraction instructions exactly.
Return only the requested JSON output.
""".strip()


EXTRACTION_TEMPLATE = """
Extract ALL diseases, disorders, syndromes, symptoms, or clinical conditions
for which the drug is explicitly indicated in the following
Indications and Usage text.

Rules:

1. Read the entire supplied text.
2. Include every explicitly stated clinical indication.
3. Include symptoms or clinical conditions when explicitly indicated
   for treatment, prevention, management, or relief.
4. Preserve clinically meaningful disease subtypes and modifiers.
5. Use wording from the supplied text as closely as possible.
6. Do not intentionally normalize or generalize disease names.
7. Do not infer conditions that are not explicitly supported by the text.
8. Exclude conditions explicitly stated as NOT indicated or NOT recommended.
9. Exclude causative organisms when they merely describe the cause of infection.
10. Exclude mechanisms, procedures, dosing information,
    treatment regimens, and patient demographics.
11. Remove duplicate clinical conditions.

Return ONLY valid JSON:

{{"diseases":["condition 1","condition 2"]}}

If there are genuinely no indications:

{{"diseases":[]}}

INDICATIONS AND USAGE:

{indication_text}
""".strip()
````

````python
def parse_output(raw_output):

    raw_output = raw_output.strip()

    try:
        parsed = json.loads(raw_output)

    except json.JSONDecodeError:

        start = raw_output.find("{")
        end = raw_output.rfind("}")

        if start == -1 or end == -1:
            raise ValueError("No JSON object found in model response.")

        parsed = json.loads(
            raw_output[start:end + 1]
        )

    if "diseases" not in parsed:
        raise ValueError("JSON does not contain 'diseases'.")

    diseases = parsed["diseases"]

    if not isinstance(diseases, list):
        raise ValueError("'diseases' must be a list.")

    cleaned = []
    seen = set()

    for disease in diseases:

        if not isinstance(disease, str):
            continue

        disease = disease.strip()

        if not disease:
            continue

        key = disease.lower()

        if key not in seen:
            seen.add(key)
            cleaned.append(disease)

    return cleaned
````

````python
def extract_openbiollm(indication_text):

    user_message = EXTRACTION_TEMPLATE.format(
        indication_text=indication_text
    )

    prompt = (
        "<|begin_of_text|>"
        "<|start_header_id|>system<|end_header_id|>\n\n"
        + SYSTEM_MESSAGE
        + "<|eot_id|>"
        "<|start_header_id|>user<|end_header_id|>\n\n"
        + user_message
        + "<|eot_id|>"
        "<|start_header_id|>assistant<|end_header_id|>\n\n"
    )

    payload = {
        "model": MODEL,
        "prompt": prompt,
        "raw": True,
        "stream": False,

        "options": {
            "temperature": 0.1,
            "num_ctx": 4096,
            "num_predict": 1000,

            "stop": [
                "<|eot_id|>",
                "<|end_of_text|>"
            ]
        }
    }

    start_time = time.time()

    response = requests.post(
        URL,
        json=payload,
        timeout=240
    )

    response.raise_for_status()

    raw_output = response.json()["response"]

    diseases = parse_output(raw_output)

    runtime = round(
        time.time() - start_time,
        2
    )

    return diseases, raw_output, runtime
````

````python
for column in [
    "openbiollm_output",
    "openbiollm_raw",
    "openbiollm_status",
    "openbiollm_runtime_sec"
]:
    if column not in df.columns:
        df[column] = ""


for i in range(5):

    record_id = df.loc[i, "record_id"]
    indication_text = str(
        df.loc[i, "indication_text"]
    )

    print()
    print("=" * 70)
    print(f"RECORD {i + 1}/5")
    print("Record ID:", record_id)

    print()
    print("EXPERT GOLD:")
    print(
        df.loc[i, "gold_expert_mentions"]
    )

    try:

        diseases, raw, runtime = extract_openbiollm(
            indication_text
        )

        df.at[i, "openbiollm_output"] = json.dumps(
            diseases,
            ensure_ascii=False
        )

        df.at[i, "openbiollm_raw"] = raw

        df.at[i, "openbiollm_status"] = (
            "OK"
            if len(diseases) > 0
            else "OK_EMPTY"
        )

        df.at[i, "openbiollm_runtime_sec"] = runtime

        print()
        print("MODEL OUTPUT:")
        print(diseases)

        print("Runtime:", runtime, "seconds")

    except Exception as e:

        df.at[i, "openbiollm_status"] = "FAILED"
        df.at[i, "openbiollm_raw"] = str(e)

        print()
        print("ERROR:")
        print(e)


df.to_excel(
    RESULT_FILE,
    index=False
)

print()
print("Saved to:")
print(RESULT_FILE)
````

````python
def run_all_openbiollm():

    total = len(df)

    for i in range(total):

        record_id = df.loc[i, "record_id"]

        # Skip records already completed
        current_status = str(
            df.at[i, "openbiollm_status"]
        ).strip()

        if current_status in ["OK", "OK_EMPTY"]:

            print(
                f"[{i + 1}/{total}] "
                f"Record {record_id} - already completed"
            )

            continue


        indication_text = str(
            df.loc[i, "indication_text"]
        )


        print()
        print("=" * 70)
        print(
            f"[{i + 1}/{total}] "
            f"Record ID: {record_id}"
        )


        success = False
        last_error = ""


        # Try maximum 3 times if there is an error
        for attempt in range(1, 4):

            try:

                diseases, raw, runtime = extract_openbiollm(
                    indication_text
                )


                df.at[
                    i,
                    "openbiollm_output"
                ] = json.dumps(
                    diseases,
                    ensure_ascii=False
                )


                df.at[
                    i,
                    "openbiollm_raw"
                ] = raw


                df.at[
                    i,
                    "openbiollm_status"
                ] = (
                    "OK"
                    if len(diseases) > 0
                    else "OK_EMPTY"
                )


                df.at[
                    i,
                    "openbiollm_runtime_sec"
                ] = runtime


                print(
                    "Output:",
                    diseases
                )

                print(
                    "Runtime:",
                    runtime,
                    "seconds"
                )


                success = True
                break


            except Exception as e:

                last_error = str(e)

                print(
                    f"Attempt {attempt}/3 failed:",
                    e
                )

                time.sleep(2)


        if not success:

            df.at[
                i,
                "openbiollm_status"
            ] = "FAILED"

            df.at[
                i,
                "openbiollm_raw"
            ] = last_error


        # Save checkpoint every 20 records
        if (
            (i + 1) % 20 == 0
            or
            i == total - 1
        ):

            df.to_excel(
                RESULT_FILE,
                index=False
            )

            print(
                ">>> CHECKPOINT SAVED TO GOOGLE DRIVE"
            )


    # Final save
    df.to_excel(
        RESULT_FILE,
        index=False
    )


    print()
    print("=" * 70)
    print("OPENBIOLLM FINISHED")
    print("=" * 70)

    print(
        "Results saved to:"
    )

    print(
        RESULT_FILE
    )
````

````python
def extract_openbiollm_raw(indication_text):

    user_message = EXTRACTION_TEMPLATE.format(
        indication_text=indication_text
    )

    prompt = (
        "<|begin_of_text|>"
        "<|start_header_id|>system<|end_header_id|>\n\n"
        + SYSTEM_MESSAGE
        + "<|eot_id|>"
        "<|start_header_id|>user<|end_header_id|>\n\n"
        + user_message
        + "<|eot_id|>"
        "<|start_header_id|>assistant<|end_header_id|>\n\n"
    )

    payload = {
        "model": MODEL,
        "prompt": prompt,
        "raw": True,
        "stream": False,
        "options": {
            "temperature": 0.1,
            "num_ctx": 4096,
            "num_predict": 1000,
            "stop": [
                "<|eot_id|>",
                "<|end_of_text|>"
            ]
        }
    }

    start_time = time.time()

    response = requests.post(
        URL,
        json=payload,
        timeout=240
    )

    response.raise_for_status()

    raw_output = response.json()["response"]

    runtime = round(
        time.time() - start_time,
        2
    )

    return raw_output, runtime
````

````python
def parse_output_robust(raw_output):

    raw_output = raw_output.strip()

    # Remove Markdown code fences if present
    raw_output = raw_output.replace("```json", "")
    raw_output = raw_output.replace("```", "")
    raw_output = raw_output.strip()

    decoder = json.JSONDecoder()

    # ------------------------------------------------
    # Try ordinary JSON
    # ------------------------------------------------
    try:
        parsed = json.loads(raw_output)

    except json.JSONDecodeError:

        # --------------------------------------------
        # Find beginning of JSON object
        # --------------------------------------------
        start = raw_output.find("{")

        if start == -1:
            raise ValueError(
                "No JSON object found in model response."
            )

        # raw_decode reads the first valid JSON object
        # and ignores text appearing afterward
        parsed, end_position = decoder.raw_decode(
            raw_output[start:]
        )

    if "diseases" not in parsed:
        raise ValueError(
            "JSON does not contain 'diseases'."
        )

    diseases = parsed["diseases"]

    if not isinstance(diseases, list):
        raise ValueError(
            "'diseases' must be a list."
        )

    cleaned = []
    seen = set()

    for disease in diseases:

        if not isinstance(disease, str):
            continue

        disease = disease.strip()

        if not disease:
            continue

        key = disease.lower()

        if key not in seen:
            seen.add(key)
            cleaned.append(disease)

    return cleaned
````

````python
failed_indices = df.index[
    df["openbiollm_status"] == "FAILED"
].tolist()

print(
    "Failed records to retry:",
    len(failed_indices)
)


for i in failed_indices:

    record_id = df.at[i, "record_id"]
    indication_text = str(
        df.at[i, "indication_text"]
    )

    print()
    print("=" * 70)
    print("Record ID:", record_id)

    try:

        # Get actual model response FIRST
        raw, runtime = extract_openbiollm_raw(
            indication_text
        )

        # Always save the actual response
        df.at[i, "openbiollm_raw"] = raw
        df.at[i, "openbiollm_runtime_sec"] = runtime

        print()
        print("RAW MODEL RESPONSE:")
        print(raw)

        # Now try parsing
        diseases = parse_output_robust(raw)

        df.at[i, "openbiollm_output"] = json.dumps(
            diseases,
            ensure_ascii=False
        )

        df.at[i, "openbiollm_status"] = (
            "OK"
            if diseases
            else "OK_EMPTY"
        )

        print()
        print("PARSED:")
        print(diseases)

    except Exception as e:

        df.at[i, "openbiollm_status"] = "FAILED"

        print()
        print("PARSING FAILED:")
        print(e)


# Save immediately
df.to_excel(
    RESULT_FILE,
    index=False
)

print()
print("Updated results saved.")
````

````python
from json_repair import repair_json
import json


# -------------------------------------------------------
# Only repair record 17348 because its intended JSON
# output is present but syntactically malformed.
# -------------------------------------------------------

idx = df.index[
    df["record_id"] == 17348
][0]

raw = df.at[idx, "openbiollm_raw"]

try:

    repaired = repair_json(raw)

    parsed = json.loads(repaired)

    diseases = parsed["diseases"]

    # Remove duplicates while preserving order
    cleaned = []
    seen = set()

    for disease in diseases:

        disease = str(disease).strip()

        if disease and disease.lower() not in seen:

            seen.add(disease.lower())
            cleaned.append(disease)

    df.at[idx, "openbiollm_output"] = json.dumps(
        cleaned,
        ensure_ascii=False
    )

    df.at[idx, "openbiollm_status"] = "OK_REPAIRED"

    print("Record 17348 repaired:")
    print(cleaned)

except Exception as e:

    print("Repair failed:", e)
````
